```python

import numpy as np

# Global Array
arr = None


# ---------------- CREATE ARRAY ---------------- #

def create_array():
    global arr

    print("\nSelect the type of array to create:")
    print("1. 1D Array")
    print("2. 2D Array")
    print("3. 3D Array")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        n = int(input("Enter number of elements: "))
        data = list(map(int, input(f"Enter {n} elements separated by space: ").split()))
        arr = np.array(data)

    elif choice == 2:
        rows = int(input("Enter the number of rows: "))
        cols = int(input("Enter the number of columns: "))

        total = rows * cols

        data = list(map(int, input(f"Enter {total} elements for the array separated by space: ").split()))

        arr = np.array(data).reshape(rows, cols)

    elif choice == 3:
        x = int(input("Enter number of blocks: "))
        y = int(input("Enter number of rows: "))
        z = int(input("Enter number of columns: "))

        total = x * y * z

        data = list(map(int, input(f"Enter {total} elements separated by space: ").split()))

        arr = np.array(data).reshape(x, y, z)

    else:
        print("Invalid Choice")
        return

    print("\nArray created successfully:")
    print(arr)

    indexing_slicing()



# ---------------- INDEXING & SLICING ---------------- #

def indexing_slicing():
    global arr

    while True:

        print("\nChoose an operation:")
        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")

        ch = int(input("Enter your choice: "))

        if ch == 1:

            if arr.ndim == 1:
                i = int(input("Enter index: "))
                print("Element =", arr[i])

            elif arr.ndim == 2:
                r = int(input("Enter row index: "))
                c = int(input("Enter column index: "))
                print("Element =", arr[r][c])

            else:
                b = int(input("Enter block index: "))
                r = int(input("Enter row index: "))
                c = int(input("Enter column index: "))
                print("Element =", arr[b][r][c])



        elif ch == 2:

            if arr.ndim == 1:
                s = int(input("Start Index: "))
                e = int(input("End Index: "))
                print("\nSliced Array:")
                print(arr[s:e])

            elif arr.ndim == 2:

                row = input("Enter the row range (start:end): ")
                col = input("Enter the column range (start:end): ")

                rs, re = map(int, row.split(":"))
                cs, ce = map(int, col.split(":"))

                print("\nSliced Array:")
                print(arr[rs:re, cs:ce])
            else:
                print("\nEnter slicing range")

                block = input("Enter Block Range (start:end): ")
                row = input("Enter Row Range (start:end): ")
                col = input("Enter Column Range (start:end): ")

                bs, be = map(int, block.split(":"))
                rs, re = map(int, row.split(":"))
                cs, ce = map(int, col.split(":"))

                print("\nSliced 3D Array:")
                print(arr[bs:be, rs:re, cs:ce])


        elif ch == 3:
            break

        else:
            print("Invalid Choice")
# ---------------- MATHEMATICAL OPERATIONS ---------------- #

def mathematical_operations():
    global arr

    if arr is None:
        print("\nFirst create an array!")
        return

    print("\nChoose a mathematical operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = int(input("Enter your choice: "))

    total = arr.size

    data = list(map(int, input(f"\nEnter the same-size array elements ({total} elements separated by space): ").split()))

    arr2 = np.array(data).reshape(arr.shape)

    print("\nOriginal Array:")
    print(arr)

    print("\nSecond Array:")
    print(arr2)

    if choice == 1:
        result = arr + arr2
        print("\nResult of Addition:")
        print(result)

    elif choice == 2:
        result = arr - arr2
        print("\nResult of Subtraction:")
        print(result)

    elif choice == 3:
        result = arr * arr2
        print("\nResult of Multiplication:")
        print(result)

    elif choice == 4:
        result = arr / arr2
        print("\nResult of Division:")
        print(result)

    else:
        print("Invalid Choice")

# ---------------- COMBINE & SPLIT ARRAYS ---------------- #

def combine_split_arrays():
    global arr

    if arr is None:
        print("\nFirst create an array!")
        return

    print("\nChoose an option:")
    print("1. Combine Arrays")
    print("2. Split Array")

    choice = int(input("Enter your choice: "))

    # ---------- COMBINE ---------- #
    if choice == 1:

        total = arr.size

        data = list(map(int, input(f"\nEnter the elements of another array to combine ({total} elements separated by space): ").split()))

        arr2 = np.array(data).reshape(arr.shape)

        print("\nOriginal Array:")
        print(arr)

        print("\nSecond Array:")
        print(arr2)

        result = np.vstack((arr, arr2))

        print("\nCombined Array (Vertical Stack):")
        print(result)

    # ---------- SPLIT ---------- #
    elif choice == 2:

        print("\nOriginal Array:")
        print(arr)

        if arr.ndim == 1:
            parts = int(input("\nEnter number of parts: "))
            result = np.array_split(arr, parts)

        elif arr.ndim == 2:
            parts = int(input("\nEnter number of row parts: "))
            result = np.array_split(arr, parts)

        else:
            print("Split is available only for 1D and 2D arrays.")
            return

        print("\nSplit Arrays:")

        for i in range(len(result)):
            print("\nPart", i + 1)
            print(result[i])

    else:
        print("Invalid Choice")
# ---------------- SEARCH, SORT & FILTER ---------------- #

def search_sort_filter():
    global arr

    if arr is None:
        print("\nFirst create an array!")
        return

    print("\nChoose an option:")
    print("1. Search a value")
    print("2. Sort the array")
    print("3. Filter values")

    choice = int(input("Enter your choice: "))

    # -------- SEARCH -------- #
    if choice == 1:

        value = int(input("\nEnter value to search: "))

        print("\nOriginal Array:")
        print(arr)

        pos = np.where(arr == value)

        if len(pos[0]) == 0:
            print("\nValue not found!")

        else:
            print("\nValue Found!")

            if arr.ndim == 1:
                print("Index:", pos[0][0])

            elif arr.ndim == 2:
                print("Row:", pos[0][0])
                print("Column:", pos[1][0])

            else:
                print("Position:", pos)

    # -------- SORT -------- #
    elif choice == 2:

        print("\nOriginal Array:")
        print(arr)

        result = np.sort(arr)

        print("\nSorted Array:")
        print(result)

        if arr.ndim == 2:
            print("(Sorting applied row-wise.)")

    # -------- FILTER -------- #
    elif choice == 3:

        value = int(input("\nShow elements greater than: "))

        print("\nOriginal Array:")
        print(arr)

        result = arr[arr > value]

        print("\nFiltered Array:")
        print(result)

    else:
        print("Invalid Choice")
# ---------------- AGGREGATES & STATISTICS ---------------- #

def aggregates_statistics():
    global arr

    if arr is None:
        print("\nFirst create an array!")
        return

    while True:

        print("\n===== Aggregates and Statistics =====")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum")
        print("7. Maximum")
        print("8. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            print("\nArray:")
            print(arr)
            print("Sum =", np.sum(arr))

        elif choice == 2:
            print("\nArray:")
            print(arr)
            print("Mean =", np.mean(arr))

        elif choice == 3:
            print("\nArray:")
            print(arr)
            print("Median =", np.median(arr))

        elif choice == 4:
            print("\nArray:")
            print(arr)
            print("Standard Deviation =", np.std(arr))

        elif choice == 5:
            print("\nArray:")
            print(arr)
            print("Variance =", np.var(arr))

        elif choice == 6:
            print("\nArray:")
            print(arr)
            print("Minimum =", np.min(arr))

        elif choice == 7:
            print("\nArray:")
            print(arr)
            print("Maximum =", np.max(arr))

        elif choice == 8:
            break

        else:
            print("Invalid Choice!")
# ---------------- FINAL MAIN PROGRAM ---------------- #

while True:

    print("\n" + "="*50)
    print("        Welcome to the NumPy Analyzer")
    print("="*50)

    print("1. Create a NumPy Array")
    print("2. Perform Mathematical Operations")
    print("3. Combine or Split Arrays")
    print("4. Search, Sort or Filter Arrays")
    print("5. Compute Aggregates and Statistics")
    print("6. Exit")

    try:
        choice = int(input("\nEnter your choice: "))

        if choice == 1:
            create_array()

        elif choice == 2:
            mathematical_operations()

        elif choice == 3:
            combine_split_arrays()

        elif choice == 4:
            search_sort_filter()

        elif choice == 5:
            aggregates_statistics()

        elif choice == 6:
            print("\nThank you for using NumPy Analyzer!")
            print("Good Bye...")
            break

        else:
            print("\nPlease Enter a Valid Choice!")

    except ValueError:
        print("\nPlease enter only numbers.")

```
